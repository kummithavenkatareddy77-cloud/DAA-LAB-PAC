```python
# Prim's Algorithm - Minimum Spanning Tree (MST)

def prim(graph, n):
    # key[i] = minimum weight of an edge connecting vertex i
    # to the vertices already included in MST
    key = [float('inf')] * n

    # parent[i] = parent of vertex i in MST
    parent = [-1] * n

    # mst_set[i] = True if vertex i is already included in MST
    mst_set = [False] * n

    # Start from vertex 0
    key[0] = 0

    for _ in range(n):
        # Find the vertex with minimum key value
        min_key = float('inf')
        u = -1

        for v in range(n):
            if not mst_set[v] and key[v] < min_key:
                min_key = key[v]
                u = v

        # Include vertex u in MST
        mst_set[u] = True

        # Update key values of adjacent vertices
        for v in range(n):
            if graph[u][v] != 0 and not mst_set[v]:
                if graph[u][v] < key[v]:
                    key[v] = graph[u][v]
                    parent[v] = u

    # Print MST
    total_cost = 0

    print("\nMinimum Spanning Tree:")
    print("Edge\t\tWeight")

    for v in range(1, n):
        print(f"{parent[v]} - {v}\t\t{graph[parent[v]][v]}")
        total_cost += graph[parent[v]][v]

    print("\nTotal MST Cost:", total_cost)

    # Complexity
    print("\n--- Complexity ---")
    print("Time Complexity: O(V^2)")
    print("Space Complexity: O(V^2)")


# ---------------- USER INPUT ----------------

n = int(input("Enter number of vertices: "))

print("\nEnter the adjacency matrix:")
print("Enter 0 if there is no edge.")

graph = []

for i in range(n):
    row = list(map(int, input(f"Row {i}: ").split()))

    if len(row) != n:
        print("Error: Enter exactly", n, "values.")
        exit()

    graph.append(row)

# Run Prim's Algorithm
prim(graph, n)
```
