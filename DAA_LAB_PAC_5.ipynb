```python
import time

# Start measuring execution time
start_time = time.perf_counter()

# Input
n = int(input("Enter number of items: "))

weights = list(map(int, input(f"Enter weights of {n} items: ").split()))
values = list(map(int, input(f"Enter values of {n} items: ").split()))

capacity = int(input("Enter knapsack capacity: "))

# Create DP table
dp = [[0 for _ in range(capacity + 1)] for _ in range(n + 1)]

# Dynamic Programming
for i in range(1, n + 1):
    for w in range(capacity + 1):

        if weights[i - 1] <= w:
            dp[i][w] = max(
                values[i - 1] + dp[i - 1][w - weights[i - 1]],
                dp[i - 1][w]
            )
        else:
            dp[i][w] = dp[i - 1][w]

# Maximum value
max_value = dp[n][capacity]

# Find selected items
selected_items = []
w = capacity

for i in range(n, 0, -1):
    if dp[i][w] != dp[i - 1][w]:
        selected_items.append(i)
        w -= weights[i - 1]

selected_items.reverse()

# End measuring execution time
end_time = time.perf_counter()

execution_time = end_time - start_time

# Output
print("\n========== OUTPUT ==========")
print("Maximum value:", max_value)
print("Selected items:", selected_items)

print("\n========== COMPLEXITY ==========")
print("Time Complexity: O(n × W)")
print("Space Complexity: O(n × W)")

print("\nExecution Time: {:.8f} seconds".format(execution_time))
```
