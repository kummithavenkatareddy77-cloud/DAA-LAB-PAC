```python
import time

# Start execution time
start_time = time.perf_counter()

# Input number of matrices
n = int(input("Enter number of matrices: "))

# Input dimensions
print("Enter the dimensions array:")
print("For n matrices, enter n+1 dimensions.")

p = list(map(int, input().split()))

# Check dimensions
if len(p) != n + 1:
    print("Error: Please enter exactly", n + 1, "dimensions.")
    exit()

# Create DP table
dp = [[0 for _ in range(n)] for _ in range(n)]

# Create table for storing optimal split positions
split = [[0 for _ in range(n)] for _ in range(n)]

# Chain length
for length in range(2, n + 1):

    for i in range(n - length + 1):

        j = i + length - 1

        dp[i][j] = float('inf')

        # Try every possible split
        for k in range(i, j):

            cost = (
                dp[i][k]
                + dp[k + 1][j]
                + p[i] * p[k + 1] * p[j + 1]
            )

            if cost < dp[i][j]:
                dp[i][j] = cost
                split[i][j] = k

# Function to print optimal parenthesization
def print_order(i, j):

    if i == j:
        return f"A{i + 1}"

    k = split[i][j]

    left = print_order(i, k)
    right = print_order(k + 1, j)

    return f"({left} × {right})"


# Minimum multiplication cost
minimum_cost = dp[0][n - 1]

# Optimal order
optimal_order = print_order(0, n - 1)

# End execution time
end_time = time.perf_counter()

execution_time = end_time - start_time

# Output
print("\n========== OUTPUT ==========")
print("Minimum number of scalar multiplications:", minimum_cost)
print("Optimal multiplication order:", optimal_order)

print("\n========== COMPLEXITY ==========")
print("Time Complexity: O(n^3)")
print("Space Complexity: O(n^2)")

print("\nExecution Time: {:.8f} seconds".format(execution_time))
```
