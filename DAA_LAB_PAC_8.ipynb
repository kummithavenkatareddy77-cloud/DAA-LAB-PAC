```python
import time
from collections import deque

# Create graph using adjacency list
graph = {}

# Input number of vertices
n = int(input("Enter number of vertices: "))

# Initialize graph
for i in range(n):
    graph[i] = []

# Input number of edges
e = int(input("Enter number of edges: "))

print("Enter edges (source destination):")

for _ in range(e):
    u, v = map(int, input().split())

    # Undirected graph
    graph[u].append(v)
    graph[v].append(u)

# Display graph
print("\nGraph:")
for vertex in graph:
    print(vertex, "->", graph[vertex])


# ---------------- DFS ----------------

def dfs(graph, start):
    visited = set()
    result = []

    def dfs_recursive(vertex):
        visited.add(vertex)
        result.append(vertex)

        for neighbor in graph[vertex]:
            if neighbor not in visited:
                dfs_recursive(neighbor)

    dfs_recursive(start)

    return result


# ---------------- BFS ----------------

def bfs(graph, start):
    visited = set()
    queue = deque()

    result = []

    visited.add(start)
    queue.append(start)

    while queue:
        vertex = queue.popleft()
        result.append(vertex)

        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return result


# Starting vertex
start_vertex = int(input("\nEnter starting vertex: "))


# DFS Execution Time
dfs_start = time.perf_counter()
dfs_result = dfs(graph, start_vertex)
dfs_end = time.perf_counter()

dfs_time = dfs_end - dfs_start


# BFS Execution Time
bfs_start = time.perf_counter()
bfs_result = bfs(graph, start_vertex)
bfs_end = time.perf_counter()

bfs_time = bfs_end - bfs_start


# ---------------- Output ----------------

print("\n========== OUTPUT ==========")

print("DFS Traversal:", " -> ".join(map(str, dfs_result)))
print("BFS Traversal:", " -> ".join(map(str, bfs_result)))

print("\n========== COMPLEXITY ==========")

print("DFS Time Complexity: O(V + E)")
print("DFS Space Complexity: O(V)")

print("BFS Time Complexity: O(V + E)")
print("BFS Space Complexity: O(V)")

print("\n========== EXECUTION TIME ==========")

print("DFS Execution Time: {:.8f} seconds".format(dfs_time))
print("BFS Execution Time: {:.8f} seconds".format(bfs_time))
```
