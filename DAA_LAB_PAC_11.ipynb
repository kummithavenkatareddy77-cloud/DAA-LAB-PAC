```python

# Floyd-Warshall Algorithm
# To find shortest paths between all pairs of vertices

# Time Complexity: O(V^3)
# Space Complexity: O(V^2)

INF = 99999

# Input number of vertices
n = int(input("Enter number of vertices: "))

# Input adjacency matrix
print("Enter the adjacency matrix:")
print("Enter 99999 for infinity (no edge).")

dist = []

for i in range(n):
    row = list(map(int, input().split()))
    dist.append(row)

# Floyd-Warshall Algorithm
for k in range(n):
    for i in range(n):
        for j in range(n):
            dist[i][j] = min(dist[i][j],
                             dist[i][k] + dist[k][j])

# Display shortest distance matrix
print("\nShortest Distance Matrix:")

for i in range(n):
    for j in range(n):
        if dist[i][j] == INF:
            print("INF", end="\t")
        else:
            print(dist[i][j], end="\t")
    print()

# Display complexities
print("\nTime Complexity: O(V^3)")
print("Space Complexity: O(V^2)")
```
