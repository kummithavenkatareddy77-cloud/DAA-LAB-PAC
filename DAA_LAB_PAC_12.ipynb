```python
import time

def tsp(graph, n):
    dp = {}

    def solve(mask, pos):
        if mask == (1 << n) - 1:
            return graph[pos][0]

        if (mask, pos) in dp:
            return dp[(mask, pos)]

        ans = float('inf')

        for city in range(n):
            if not (mask & (1 << city)):
                cost = graph[pos][city] + solve(mask | (1 << city), city)
                ans = min(ans, cost)

        dp[(mask, pos)] = ans
        return ans

    result = solve(1, 0)
    return result, len(dp)


# User Input
n = int(input("Enter number of cities: "))

print("Enter the distance matrix:")
graph = []

for i in range(n):
    row = list(map(int, input().split()))
    graph.append(row)

# Measure execution time
start_time = time.perf_counter()

minimum_cost, states = tsp(graph, n)

end_time = time.perf_counter()

# Display Results
print("\nMinimum travelling cost:", minimum_cost)
print("Time Complexity: O(n^2 * 2^n)")
print("Space Complexity: O(n * 2^n)")
print("Execution Time:", end_time - start_time, "seconds")
```
